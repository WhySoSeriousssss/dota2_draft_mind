```python
import requests
import json

open_dota_url = "https://api.opendota.com/api/explorer"
```

```python
def explorer(sql):
    response = requests.get(
        open_dota_url,
        params={"sql": sql},
        timeout=90,
    )

    response.raise_for_status()

    data = response.json()

    if "error" in data:
        raise RuntimeError(data["error"])

    return data["rows"]
```

```python
url = "https://api.opendota.com/api/matches/8963045259"
ret = requests.get(url).json()

print(ret)
```

```python
sql = """
SELECT *
FROM public_matches
WHERE
    lobby_type = 7
ORDER BY start_time DESC
LIMIT 10000
"""
rows = explorer(sql)

with open("public_matches.json", "w") as f:
    json.dump(rows, f, indent=4)
```

```python
url = "https://api.opendota.com/api/constants/heroes"

heroes = requests.get(url).json()

with open("heroes.json", "w") as f:
    json.dump(heroes, f, indent=4)
```

```python
sql = """
SELECT COUNT(*)
FROM public_matches
WHERE
    start_time >= 1785600000
    AND start_time < 1787673600
    AND lobby_type = 7
    AND game_mode = 22;
"""
rows = explorer(sql)

print(rows)
```
